```
import numpy as np
import qutip
```

```
# ---------------------------------------------------------
# Step 1: Choose the two angles that describe the qubit.
# theta_angle = angle measured down from the north pole (z axis)
# psi_angle   = angle measured around the equator from the x axis
#               (many textbooks call this angle "phi")
# ---------------------------------------------------------
theta_angle = np.pi / 3
psi_angle = np.pi / 4
```

```
# ---------------------------------------------------------
# Step 2: Build the qubit state from the two angles.
# state = cos(theta/2)|0> + e^(i*psi) sin(theta/2)|1>
# ---------------------------------------------------------
amplitude_of_zero = np.cos(theta_angle / 2)
amplitude_of_one = np.exp(1j * psi_angle) * np.sin(theta_angle / 2)
state_before_hadamard = np.array([amplitude_of_zero, amplitude_of_one])
```

```
# ---------------------------------------------------------
# Step 3: Write down the Hadamard gate as a 2x2 matrix.
# ---------------------------------------------------------
hadamard_gate = (1 / np.sqrt(2)) * np.array([[1, 1],
                                             [1, -1]])
```

```
# ---------------------------------------------------------
# Step 4: Apply the Hadamard gate (matrix times vector).
# ---------------------------------------------------------
state_after_hadamard = hadamard_gate @ state_before_hadamard
```

```
# ---------------------------------------------------------
# Step 5: Turn a qubit state into a point (x, y, z) on the Bloch sphere.
# x = 2 * Real( conj(a) * b )
# y = 2 * Imag( conj(a) * b )
# z = |a|^2 - |b|^2
# where a = amplitude of |0>, b = amplitude of |1>
# ---------------------------------------------------------
def convert_state_to_bloch_vector(state):
    a = state[0]
    b = state[1]
    x = 2 * np.real(np.conj(a) * b)
    y = 2 * np.imag(np.conj(a) * b)
    z = abs(a) ** 2 - abs(b) ** 2
    return [x, y, z]

bloch_vector_before = convert_state_to_bloch_vector(state_before_hadamard)
bloch_vector_after = convert_state_to_bloch_vector(state_after_hadamard)

print("Bloch vector before Hadamard (x, y, z):", np.round(bloch_vector_before, 3))
print("Bloch vector after  Hadamard (x, y, z):", np.round(bloch_vector_after, 3))
```

```
# ---------------------------------------------------------
# Step 6: Draw the Bloch sphere BEFORE the Hadamard gate (ready-made QuTiP sphere).
# ---------------------------------------------------------
bloch_sphere_before = qutip.Bloch()
bloch_sphere_before.add_vectors(bloch_vector_before)
bloch_sphere_before.vector_color = ["blue"]
bloch_sphere_before.show()
```

```
# ---------------------------------------------------------
# Step 7: Draw the Bloch sphere AFTER the Hadamard gate (ready-made QuTiP sphere).
# ---------------------------------------------------------
bloch_sphere_after = qutip.Bloch()
bloch_sphere_after.add_vectors(bloch_vector_after)
bloch_sphere_after.vector_color = ["red"]
bloch_sphere_after.show()
```
